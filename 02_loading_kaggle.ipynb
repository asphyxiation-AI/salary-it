# Загрузка датасетов hh.ru с Kaggle

## Описание
Этот блокнот скачивает два датасета с hh.ru через `kagglehub`:
1. **hh IT 2026** (6 968 IT-вакансий, 42 признака).
2. **hh 2026** (129 464 вакансии, 38 признаков; фильтруем до IT-категории).

## Требования
- Аккаунт Kaggle.
- API-токен: https://www.kaggle.com/settings → API → Create New Token.
- В Colab: слева 🔑 → Add secret → `KAGGLE_KEY`.

## Результат
- `data/hh_it_vacancies.csv` — 6 968 строк.
- `data/hh_vacancies_2026_07.csv` — 129 464 строки (не в репозитории, слишком большой).

```python
!pip install -q kagglehub[pandas-datasets]

import os
from google.colab import userdata

# Настройка Kaggle API (через Colab secrets)
os.environ["KAGGLE_KEY"] = userdata.get("KAGGLE_KEY")

import kagglehub
import shutil
os.makedirs("data", exist_ok=True)
```

## Датасет 1: hh IT 2026 (skeethook)

6 968 уникальных IT-вакансий, 42 признака: работодатель, рейтинг, город, зарплата, опыт, формат работы, дата публикации, число откликов.

```python
import glob

path = kagglehub.dataset_download("skeethook/hh-ru-it-vacancies-russia-2026")
print(f"Скачано в: {path}")

# Копируем в data/
hh_it_files = glob.glob(f"{path}/**/*.csv", recursive=True)
print(f"Найдены файлы: {hh_it_files}")

if hh_it_files:
    shutil.copy(hh_it_files[0], "data/hh_it_vacancies.csv")
    print(f"✓ Скопировано: data/hh_it_vacancies.csv")

    import pandas as pd
    df_hh_it = pd.read_csv("data/hh_it_vacancies.csv")
    print(f"Размер: {df_hh_it.shape}")
    print(f"Колонки: {list(df_hh_it.columns)}")
```

## Датасет 2: hh 2026 (anastasiavdovina)

129 464 вакансии по 27 профессиональным полям. Фильтруем до IT-категории (~5 850).

⚠️ **Файл ~100 МБ** — не загружаем в репозиторий, скачиваем при каждом запуске.

```python
path = kagglehub.dataset_download("anastasiavdovina/russian-job-market-vacancies-from-hh-ru-2026")
print(f"Скачано в: {path}")

hh_2026_files = glob.glob(f"{path}/**/hh_vacancies_2026_07.csv", recursive=True)
print(f"Найдены файлы: {hh_2026_files}")

if hh_2026_files:
    shutil.copy(hh_2026_files[0], "data/hh_vacancies_2026_07.csv")
    print(f"✓ Скопировано: data/hh_vacancies_2026_07.csv")

    import pandas as pd
    df_hh_2026 = pd.read_csv("data/hh_vacancies_2026_07.csv")
    print(f"Размер: {df_hh_2026.shape}")
    print(f"Колонки: {list(df_hh_2026.columns)}")
```
