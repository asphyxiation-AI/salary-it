Установка датасетов (trudvsem - github)

```python
import os
import shutil
import glob

os.makedirs("data", exist_ok=True)

# 1. trudvsem — берём последний
trud_files = sorted(glob.glob("vacancies_full_*.csv"))
if trud_files:
    shutil.copy(trud_files[-1], "data/vacancies_trudvsem.csv")
    print(f"✓ trudvsem: {trud_files[-1]} → data/vacancies_trudvsem.csv")

# 2. hh IT 2026
hh_it_paths = glob.glob("/root/.cache/kagglehub/datasets/skeethook/hh-ru-it-vacancies-russia-2026/**/*.csv", recursive=True)
if hh_it_paths:
    shutil.copy(hh_it_paths[0], "data/hh_it_vacancies.csv")
    print(f"✓ hh IT: {hh_it_paths[0]} → data/hh_it_vacancies.csv")

# 3. hh 2026 — берём только основной файл, не data_dictionary
hh_2026_paths = glob.glob("/root/.cache/kagglehub/datasets/anastasiavdovina/russian-job-market-vacancies-from-hh-ru-2026/**/hh_vacancies_2026_07.csv", recursive=True)
if hh_2026_paths:
    shutil.copy(hh_2026_paths[0], "data/hh_vacancies_2026_07.csv")
    print(f"✓ hh 2026: {hh_2026_paths[0]} → data/hh_vacancies_2026_07.csv")

# Проверка
print("\nСодержимое data/:")
for f in sorted(os.listdir("data")):
    size_mb = os.path.getsize(f"data/{f}") / (1024*1024)
    print(f"  {f} — {size_mb:.1f} MB")
```
